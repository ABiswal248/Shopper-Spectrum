### Conclusion

This project, Shopper Spectrum: Customer Segmentation and Product Recommendation System, successfully analyzed e-commerce transaction data to understand customer purchasing behavior and improve customer experience.

Through data cleaning, feature engineering, and exploratory data analysis, meaningful insights were extracted from the dataset. Using RFM Analysis (Recency, Frequency, Monetary) and K-Means Clustering, customers were segmented into four categories:

High Value Customers
Regular Customers
Occasional Customers
At-Risk Customers

These segments help businesses design targeted marketing campaigns, improve customer retention, and maximize revenue.

In addition, an Item-Based Collaborative Filtering Recommendation System was developed using Cosine Similarity to recommend products based on customer purchase patterns. This recommendation mechanism works similarly to those used by major e-commerce platforms such as Amazon and Flipkart.

Finally, an interactive Streamlit Web Application was created to provide real-time customer segmentation and product recommendations through a user-friendly interface.
